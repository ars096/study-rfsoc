# proj018 — specd の中身の確かめ（C-1〜C-5、実機・SG）

`s45fcheck.py` の冒頭に各試験の予言と合否。**各セルの前に書いた入力の繋ぎ方にしてから実行する**。

SG・減衰器・分配器の型番と減衰量を、このノートの終わりに書いておく（README に写す）。

```python
from s45client import S45
import s45fcheck as F
HOST = "<board>"      # ボードの IP
F_SG = 3000.5          # MHz
s = S45(HOST)
s.status()
```

## C-1 周波数軸の向き
SG の CW（F_SG）を **ADC_A** に入れる（分配器のままで可）。1 回目の後、SG を **+1 MHz** 動かして 2 回目。

```python
r1 = F.c1(s, F_SG, adc="A")
```

```python
# SG を F_SG + 1 MHz にしてから
r2 = F.c1(s, F_SG + 1.0, adc="A", ref=r1)
```

## C-2 設定が効く
入力はそのままでよい（中身ではなく、設定と記録の頭・周波数軸を見る）。

```python
F.c2(s)
```

## C-3 ADC と窓の対応
SG を F_SG に戻し、**1 本の ADC だけ**に入れる（分配器を外す）。下の adc をその ADC に。ほかの 3 本は 50 Ω で終端か、何も繋がない。

```python
F.c3(s, F_SG, adc="A")
```

## C-4 SHIFT
SG を ADC_A に。山が SHIFT 9 で飽和しないレベル（−20 dBm 前後 ＋ 減衰）で。

```python
F.c4(s, F_SG, adc="A")
```

## C-5 時刻
入力はそのままでよい。

```python
F.c5(s)
```
